```python
# Import necessary libraries
import nltk
from nltk.corpus import stopwords

# Download necessary resources for nltk
nltk.download('punkt')
nltk.download('stopwords')

# Load English stopwords
stop_words = set(stopwords.words('english'))

# Read the content of the file
file_path = "Data_1.txt"  # Ensure the file is in the same directory as the notebook
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

print("Original Text:\n", text)

from nltk.tokenize import word_tokenize

# Tokenization using NLTK
nltk_tokens = word_tokenize(text.lower())  # Convert to lowercase and tokenize
nltk_tokens_filtered = [word for word in nltk_tokens if word.isalnum() and word not in stop_words]  # Remove punctuation & stop words

print("\nTokenization using NLTK:\n", nltk_tokens_filtered)
```

```python
import spacy
import string

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

# Read the file directly
file_path = "Data_1.txt"
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Process text using spaCy
doc = nlp(text.lower())  # Convert to lowercase and process text

# Extract stop words found in the text
stop_words_found = {token.text for token in doc if token.is_stop}

# Tokenization (excluding stopwords and punctuation)
filtered_tokens = [token.text for token in doc if token.is_alpha and not token.is_stop]

# Join tokens back into a cleaned sentence
cleaned_text = " ".join(filtered_tokens)

# Display Results
print("\nOriginal Text:\n", text)
print("\nCleaned Text (After Stop Words and Punctuation Removal):\n", cleaned_text)
print("\nStop Words Found in the Given Text Corpus:\n", stop_words_found)
```

```python
# Import necessary libraries
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download necessary resources for nltk
nltk.download('punkt')
nltk.download('stopwords')

# Load English stopwords
stop_words = set(stopwords.words('english'))

# Read the content of the file
file_path = "Data_1.txt"  # Ensure the file is in the same directory as the notebook
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

print("Original Text:\n", text)

# Tokenization using NLTK
nltk_tokens = word_tokenize(text.lower())  # Convert to lowercase and tokenize

# Identify stop words found in the text
stop_words_found = {word for word in nltk_tokens if word in stop_words}

# Remove punctuation and stop words
nltk_tokens_filtered = [word for word in nltk_tokens if word.isalnum() and word not in stop_words]

# Join tokens into a cleaned sentence
cleaned_text = " ".join(nltk_tokens_filtered)

# Display Results
print("\nCleaned Text (After Stop Words and Punctuation Removal):\n", cleaned_text)
print("\nStop Words Found in the Given Text Corpus:\n", stop_words_found)
```
